```python
# Llamado de modulos de PySpark para esta sesión

from pyspark.sql import SparkSession
from pyspark.sql.functions import *
```

```python
# Crear sesión de Spark

spark = (SparkSession
         .builder
         .appName("Esquemas")
         .getOrCreate())
```

```python
# Esquema inferido

Flight = spark.read.csv(
    "/content/drive/MyDrive/flight data.csv",
    header = True,
    inferSchema=True
)
```

```python
Flight.printSchema()
```

```python
Flight.show(truncate=False)
```

```python
# Esquema explícito

from pyspark.sql.types import *

esquema = StructType([
    StructField("from_airport_code", StringType(), True),
    StructField("from_country", StringType(), True),
    StructField("dest_airport_code", StringType(), True),
    StructField("dest_country", StringType(), True),
    StructField("aircraft_type", StringType(), True),
    StructField("airline_number", StringType(), True),
    StructField("airline_name", StringType(),  True),
    StructField("flight_number", StringType(), True),
    StructField("departure_time", TimestampType(), True),
    StructField("arrival_time", TimestampType(), True),
    StructField("duration", IntegerType(),  True),
    StructField("stops", IntegerType(), True),
    StructField("price", DoubleType(), True),
    StructField("currency", StringType(), True),
    StructField("co2_emission", DoubleType(), True),
    StructField("avg_co2_emission_for_this_route", DoubleType(), True),
    StructField("co2_percentage", StringType(), True),
    StructField("scan_date", TimestampType(), True)
])

Flight_esquema = spark.read.csv(
    "/content/drive/MyDrive/flight data.csv",
    header = True,
    schema = esquema
).withColumn(
        "co2_percentage",
        regexp_replace("co2_percentage", "%", "").cast("double")/100
    )
```

```python
Flight_esquema.printSchema()
```

```python
Flight_esquema.show(truncate=False)
```
