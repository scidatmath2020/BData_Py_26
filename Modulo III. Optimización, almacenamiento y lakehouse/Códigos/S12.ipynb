```python
# Llamado de modulos de PySpark para la sesión

from pyspark.sql import SparkSession
from pyspark.sql.functions import *
```

```python
# Crear sesión de Spark

spark = (SparkSession
         .builder
         .appName("Lectura_Archivos")
         .getOrCreate())
```

```python
# Lectura de CSV

Flight_csv = spark.read.csv(
    "/content/drive/MyDrive/flight data.csv",
    header = True,
    inferSchema = True
)

Flight_csv.count()
```

```python
Flight_csv.show()
```

```python
# Lectura parquet

Flight_parquet = spark.read.parquet(
    "/content/drive/MyDrive/flight data.parquet"
)

Flight_parquet.printSchema()
```

```python
Flight_parquet.count()
```

```python
Flight_parquet.count()
```

```python
# Lectura JSON

Flight_JSON = spark.read.json(
    "/content/drive/MyDrive/flight data.json"
)

Flight_JSON.printSchema()
```

```python
Flight_JSON.show()
```

```python
# Lectura ORC

Flight_ORC = spark.read.orc(
    "/content/drive/MyDrive/flight data.orc"
)

Flight_ORC.printSchema()
```

```python
Flight_ORC.count()
```

```python
Flight_ORC.show()
```
